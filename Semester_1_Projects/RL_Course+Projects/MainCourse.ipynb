# Load the Environment

```python
env_name = "CartPole-v0"
env = gym.make(env_name, render_mode="human")
```

```python
episodes = 5
for episode in range(1, episodes+1):

    observation, info = env.reset() # This resets the environment everytime

    print(f"Starting Observation: {observation}\nStarting Info: {info}")

    done = False
    score = 0 

    while not done:
        env.render() # Renders the Environment for us to see
        action = env.action_space.sample() # Generates a random action sample from cases provided by the probability policy
        observation, reward, terminated, truncated, info = env.step(action) # Then we pass the action 
        score += reward # Update the Reward
        done = terminated or truncated # Check if episode is complete 
    
    print(f"Episode: {episode}\nObervation: {observation}\nInfo: {info}\nReward: {reward} Score: {score}\n")
env.close() # This closes the environment
```

# Training an RL Model

```python
# Creating Folder for logs and saved models

LOG_PATH = "Logs/"
SAVED_MODELS_PATH = "Saved Models/"

os.makedirs(LOG_PATH, exist_ok=True)
os.makedirs(SAVED_MODELS_PATH, exist_ok=True)

```

```python
env = gym.make(env_name)
env = DummyVecEnv([lambda: env])
model = PPO("MlpPolicy", env, verbose=1, tensorboard_log=LOG_PATH)
```

```python
model.learn(total_timesteps=20000)
```

# Saving and Reloading the Model

```python
model.save(SAVED_MODELS_PATH+"/PPO_CartPole")

del model
```

```python
model  = PPO.load(SAVED_MODELS_PATH+"/PPO_CartPole", env=env)
```
