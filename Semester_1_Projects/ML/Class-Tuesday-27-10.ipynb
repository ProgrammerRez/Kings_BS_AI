```python
import pandas as pd
import numpy as np

data = np.array([
    [2, 60, "Campus", 48],
    [4, 75, "Online", 60],
    [None, 82, "Campus", 85],
    [7, 105, "Online", 80],
    [5, 80, "Campus", 72]
])

columns = ["Hours", "Attendance", "Mode", "Score"]

df = pd.DataFrame(data=data, columns=columns)
df.head()
```

```python
print(f"\nShape: {df.shape}")
print(f"\nMissing Values: {df.isna().sum()}")
print(f"\nOriginal Data: {df}")
```

```python
df.to_csv("students.csv", index=False)
```
